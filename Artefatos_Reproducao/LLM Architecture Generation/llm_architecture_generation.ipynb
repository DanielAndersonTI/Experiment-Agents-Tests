```python
# ========================================================
# Configuration Management - OpenAI Integration
# ========================================================
#
# This script loads the OpenAI API key from a .env file
# and defines utility functions and prompts to query the
# OpenAI API using a chat-completion model.
#
# === API Key ===
# The API key is expected to follow the format:
#   -> sk-...zz8A
#   Note: The key is unique to each user/account/project.
#
# === Model Settings ===
# Default model: "o3"
#
# === Output Files ===
# Results are saved in CSV format in the Results directory.
# Ensure the directories exist before running this script.
# ==========================================================
```

```python
# All imports

from dotenv import load_dotenv
import os
from openai import OpenAI
```

```python
# Load API key from .env file
load_dotenv()
openai_api_key = os.getenv("OPENAI_API_KEY")
client = OpenAI(api_key=openai_api_key)
```

```python
# ========================================================================
# Function: run_prompt
# Description: Executes a prompt using the configured model.
# Parameters:
#   - prompt: string containing the task description.
#   - model: OpenAI model to use (default: "o3").
# Returns:
#   - The content of the first response message, stripped of whitespace.
# =======================================================================
def run_prompt(prompt, model="o3", temperature=0.3):
    response = client.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": prompt}],
    )
    return response.choices[0].message.content.strip()
```

## Spring PetClinic – Zero-Shot

```python
prompt = """Given the following requirements:
Spring PetClinic Microservices - Requirements

1. Manage Clients:
   - Add, update, and delete client information.
   - View detailed information about existing clients.

2. Manage Pets:
   - Add, update, and delete pet information.
   - Link pets to their respective owners.

3. Manage Visits:
   - Register new visits for pets.
   - View visit history for a specific pet.

4. Manage Veterinarians:
   - Add, update, and delete vet information.
   - View specialties for each veterinarian.

5. Service Discovery:
   - Register and locate services using Eureka.

6. Centralized Configuration:
   - Manage service configurations through a Config Server.

7. API Gateway:
   - Route client requests to appropriate backend services.

8. Monitoring and Administration:
   - Monitor services and applications using Admin Server.

Generate a list of microservices in CSV format with the following columns:
- Microservice
- Responsibilities (separate multiple responsibilities with semicolons)
- Communicates With (separate multiple services with semicolons)

Output only the CSV data without explanations.
"""
result = run_prompt(prompt)
with open('Results/Zero-Shot Results/petclinic-zerozhot-results.csv', 'w', encoding='utf-8') as f:
    f.write(result)
```
